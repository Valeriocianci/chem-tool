```python
#Calcolo singolo densità

massa_oro = 38.64
volume_oro = 2.0
densità = massa_oro / volume_oro
print("la densità calcolata in g/cm^3")
print(densità)
```

```python
#Macchina addestrata per densità
def calcola_densità(massa, volume):
    risultato = massa / volume
    return risultato
densità_bulk = 38.64, 2.0
densità_nanostars= 0.015, 0.0007
print("densità oro bulk", densità_bulk)
print("densità oro nanostars", densità_nanostars)
```

```python
#messa in sicurezza calcolo densità
def calcola_densità(massa,volume):
    if volume == 0 :
       print("attenzione rilevato un volume pari a 0. Calcolo annullato")
       return 0
    else:
       risultato = massa / volume
       return risultato
prova_errore = calcola_densità(15.0, 0)
```

```python
#elaborate a lot of data
masse_campioni = [38.64, 40.12, 35.50, 0]
volume_standard = 2.0
print("inizio l'analisi dei campioni...\n")
for massa_attuale in masse_campioni :
  dens = calcola_densità(massa_attuale, volume_standard)
  print("Massa", massa_attuale, "->Densità", dens)
  print("\nAnalisi completata!")
```
