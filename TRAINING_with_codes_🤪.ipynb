2) Code for molar concentration calculation

```python
#data
mass_grams= 5.85
molar_mass= 58.5
volume_liters= 0.25

#number of moles
moles= mass_grams / molar_mass

#molarity
molarity= moles / volume_liters

print("Calculated molarity:")
print(f"Value: {molarity: .2f}")

#evaluate the concentration
if molarity>= 0.5:
  print(" this is a high concentration")
else:
  print("this is a low concentration")
```

3) code for the following simulation: i have completed four different syntheses of AuNSs; i have the actual mass obtained for each batch and i need to calculate the percentage yield based on a fixed theoretical mass. If the yield is 75% higher, it is acceptable; otherwise, it is a failed batch

```python
#actual masses obtained from 4 different syntheses ( in mg)
actual_mass= [38.5, 42.1, 25.0, 48.2]
theoretical_mass= 50
print("---synthesis batch analysis")

#1. i'm setting up the loop to extract one mass at a time
for mass in actual_mass:
  #2. now i'm calculating the percentage yield:
  yield_percentage= (mass/theoretical_mass)*100
  #3. print the yield formatted to exactly 1 decimal place
  print(f"Calculated Yield: {yield_percentage:.1f}")
  #4. evaluate the yield with an if/else block:
  if yield_percentage >= 75.0 :
    print("status: Acceptable")
  else:
    print("status:Failed- Review protocol")
```

Introducing new tool .append(): the scenario is the following--> i'm monitoring a flow reactor; i have a list of recorded temperautres and the safe operating limit is 150.0°C; i need to iterate throgh the data, print a warning for any temperature tha exceeds this limit, and save only the safe temperatures into a brand new list

```python
#raw data from the reactor
recorded_temps= [145.2, 160.5,148.9, 155.0, 149.5]

#empty list, ready to store our filtered data
safe_temps= []
print("---Temperature Check---")

#1. Set up the loop
for tempt in recorded_temps:
  #2.check if the temperature is strictly greater than 150.0
  if tempt > 150.0:
    print(f"warning!!! {tempt:.1f} °C is too high")

  #3.handle the safe temperature
  else:
    print("perfect condition")
    #4. save the safe temperature into the new list using .append()
    safe_temps.append(tempt)
#printing the final condition
print("\nFinal safe temperature list:")
print(safe_temps)
```
